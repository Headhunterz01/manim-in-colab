# Sample1:Gravity Assist

```python
%%manim -qm -v WARNING GravityAssist

from manim import *
import numpy as np


class GravityAssist(Scene):
    def construct(self):
        # =========================
        # 1. Background
        # =========================
        self.camera.background_color = "#050816"

        # Title
        title = Text(
            "GRAVITY ASSIST",
            font_size=38,
            weight=BOLD
        ).to_edge(UP)

        subtitle = Text(
            "引力弹弓效应",
            font_size=22
        ).next_to(title, DOWN, buff=0.08)

        self.play(
            FadeIn(title, shift=DOWN * 0.2),
            FadeIn(subtitle)
        )

        # =========================
        # 2. Planet
        # =========================
        planet_pos = np.array([2.0, 0.0, 0.0])

        planet = Circle(
            radius=0.72,
            fill_opacity=1,
            stroke_width=3
        ).move_to(planet_pos)

        # Atmospheric glow
        glow1 = Circle(
            radius=0.92,
            stroke_width=2,
            stroke_opacity=0.35
        ).move_to(planet_pos)

        glow2 = Circle(
            radius=1.15,
            stroke_width=2,
            stroke_opacity=0.12
        ).move_to(planet_pos)

        planet_label = Text(
            "PLANET",
            font_size=20
        ).next_to(planet, DOWN, buff=0.15)

        self.play(
            GrowFromCenter(glow2),
            GrowFromCenter(glow1),
            GrowFromCenter(planet),
            FadeIn(planet_label)
        )

        # =========================
        # 3. Star field
        # =========================
        np.random.seed(7)

        stars = VGroup()

        for _ in range(70):
            x = np.random.uniform(-6.5, 6.5)
            y = np.random.uniform(-3.5, 3.5)

            # Don't overcrowd planet
            if np.linalg.norm(np.array([x, y]) - planet_pos[:2]) < 1.5:
                continue

            star = Dot(
                point=np.array([x, y, 0]),
                radius=np.random.uniform(0.008, 0.025),
                fill_opacity=np.random.uniform(0.3, 0.9)
            )
            stars.add(star)

        self.add(stars)

        # =========================
        # 4. Incoming trajectory
        # =========================
        start = np.array([-6.2, 1.55, 0])
        end = np.array([6.0, -2.0, 0])

        # A curved flyby path
        def trajectory(t):
            # t: 0 -> 1
            x = -6.2 + 12.2 * t
            y = 1.55 - 3.55 * t

            # gravitational deflection near planet
            bend = 1.65 * np.exp(-((t - 0.68) / 0.17) ** 2)

            y += bend

            return np.array([x, y, 0])

        # Sample trajectory
        points = [
            trajectory(t)
            for t in np.linspace(0, 1, 300)
        ]

        path = VMobject()
        path.set_points_smoothly(points)

        # Ghost path = what happens without gravity
        straight_path = DashedLine(
            start,
            end,
            dash_length=0.12,
            stroke_width=2,
            stroke_opacity=0.25
        )

        self.play(
            Create(straight_path),
            Create(path),
            run_time=1.8
        )

        # =========================
        # 5. Spacecraft
        # =========================
        ship = Triangle(
            fill_opacity=1,
            stroke_width=1
        ).scale(0.16)

        ship.rotate(-0.28)
        ship.move_to(trajectory(0))

        # Trail
        trail = VMobject()
        trail.set_points_as_corners([trajectory(0), trajectory(0)])

        def update_trail(mob):
            current = ship.get_center()

            old_points = mob.get_all_points()

            if len(old_points) == 0:
                mob.set_points_as_corners([current, current])
                return

            pts = list(old_points)
            pts.append(current)

            # Limit trail length
            pts = pts[-180:]

            mob.set_points_smoothly(pts)

        trail.add_updater(update_trail)

        # =========================
        # 6. Velocity vector
        # =========================
        velocity_arrow = Arrow(
            ORIGIN,
            RIGHT,
            buff=0,
            stroke_width=4,
            max_tip_length_to_length_ratio=0.2
        )

        velocity_arrow.add_updater(
            lambda a: self.update_velocity_arrow(
                a,
                ship
            )
        )

        velocity_label = Text(
            "v",
            font_size=26
        )

        velocity_label.add_updater(
            lambda m: m.next_to(
                velocity_arrow.get_end(),
                UP,
                buff=0.08
            )
        )

        self.add(trail, ship, velocity_arrow, velocity_label)

        # =========================
        # 7. Physics HUD
        # =========================
        hud = VGroup()

        speed_before = Text(
            "v₁  =  1.00",
            font_size=20
        )

        speed_after = Text(
            "v₂  =  1.38",
            font_size=20
        )

        deflection = Text(
            "Δθ  ≈  48°",
            font_size=20
        )

        hud.add(
            speed_before,
            speed_after,
            deflection
        )

        hud.arrange(
            DOWN,
            aligned_edge=LEFT,
            buff=0.08
        )

        hud.to_corner(
            UL,
            buff=0.35
        )

        hud_box = SurroundingRectangle(
            hud,
            buff=0.18,
            stroke_width=1.5
        )

        self.play(
            FadeIn(hud_box),
            FadeIn(hud)
        )

        # =========================
        # 8. Animate flyby
        # =========================
        t = ValueTracker(0)

        ship.add_updater(
            lambda m: m.move_to(
                trajectory(t.get_value())
            )
        )

        # Rotate ship according to local trajectory direction
        def rotate_ship(m):
            tt = min(t.get_value() + 0.01, 1)
            direction = trajectory(tt) - trajectory(t.get_value())

            if np.linalg.norm(direction) > 0:
                angle = np.arctan2(
                    direction[1],
                    direction[0]
                )
                m.set_angle(angle)

        ship.add_updater(rotate_ship)

        self.play(
            t.animate.set_value(1),
            run_time=7,
            rate_func=linear
        )

        # =========================
        # 9. Highlight energy transfer
        # =========================
        energy_text = Text(
            "GRAVITY CHANGES THE TRAJECTORY",
            font_size=24,
            weight=BOLD
        ).to_edge(DOWN)

        self.play(
            FadeIn(
                energy_text,
                shift=UP * 0.2
            )
        )

        # Highlight planet
        self.play(
            Indicate(
                planet,
                scale_factor=1.15,
                run_time=1.2
            )
        )

        self.wait(2)

        # Cleanup
        ship.clear_updaters()
        trail.clear_updaters()
        velocity_arrow.clear_updaters()
        velocity_label.clear_updaters()


    def update_velocity_arrow(self, arrow, ship):
        center = ship.get_center()

        # Approximate local velocity from trajectory
        # based on the current position.
        direction = RIGHT

        # Find nearest point on the trajectory
        # using x-coordinate approximation.
        x = center[0]

        if x < 6:
            dx = 0.05

            # approximate trajectory
            def traj_x(x):
                t = (x + 6.2) / 12.2
                y = 1.55 - 3.55 * t
                y += 1.65 * np.exp(
                    -((t - 0.68) / 0.17) ** 2
                )
                return np.array([x, y, 0])

            p1 = traj_x(x)
            p2 = traj_x(min(x + dx, 6))

            direction = p2 - p1

        if np.linalg.norm(direction) > 0:
            direction = direction / np.linalg.norm(direction)

        start = center
        end = center + direction * 0.65

        arrow.put_start_and_end_on(
            start,
            end
        )
```
